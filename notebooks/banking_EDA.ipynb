```python
import os
from dotenv import load_dotenv
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sqlalchemy import create_engine


load_dotenv()

HOST = os.getenv("HOST")
USER = os.getenv("USER")
PASSWORD = os.getenv("PASSWORD")
DATABASE = os.getenv("DB")
PORT = os.getenv("PORT")


try:
    engine = create_engine(f"mysql+mysqlconnector://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}")
    print(f"Engine created successfully for database: {DATABASE}")
     
except Exception as err:
    print(f"Error creating engine: {err}")
   
```

```python
query = """
SELECT
    *
FROM
    banking_db.banking;
"""
```

```python
df = pd.read_sql(query,engine)
```

### Exploratory Data Analyis (EDA)

```python
# calculate income band based on the Estimated Income and categorized as Low,Medium or High
bins = [0,100000,300000,float('inf')]
labels = ["Low","Medium","High"]

df["Income Band"] = pd.cut(df["Estimated Income"],bins=bins,labels=labels,right=False)
```

### Numerical analysis

```python
numerical_columns = ['Estimated Income',
                     'Superannuation Savings',
                     'Credit Card Balance',
                     'Bank Loans'
                     ,'Bank Deposits',
                     'Checking Accounts',
                     'Saving Accounts',
                     'Foreign Currency Account',
                     'Business Lending'

]
plt.figure(figsize=(20,20))
for i,col in enumerate(numerical_columns):
    plt.subplot(4,3,i+1)
    sns.histplot(df[col],kde=True)
    plt.title(col)
plt.show()
```

### Multivariate analysis using Heatmap


```python
numerical_corr = df[numerical_columns].corr()
plt.figure(figsize=(12,12))
sns.heatmap(numerical_corr,annot=True,cmap='crest')
plt.title("Correlation Matrix for Numerical columns")
plt.show()
```

### Insights from the EDA - Correlation Matrix

### Strong Relationships
There is a very strong positive correlation between bank deposits and checking accounts (0.84). This indicates that customers who maintain high levels of bank deposits also tend to hold higher balances in their checking accounts. These two products move closely together and represent core transactional banking behaviour.

Bank deposits also show a strong positive correlation with saving accounts (0.75). This suggests that customers with higher deposits generally exhibit strong saving behaviour across multiple account types.

### Moderate Relationships
Business lending is moderately correlated with bank loans (0.42). This implies that customers who rely on bank loans are also likely to engage in business lending activities, indicating broader credit usage.

There is a moderate positive correlation between business lending and bank deposits (0.44). This shows that customers involved in business borrowing often maintain substantial deposit balances, making them potentially high-value clients.

Credit card balance has a moderate correlation with bank loans (0.37). This suggests that customers who carry higher credit card balances are also more likely to have bank loans, reflecting greater dependence on borrowed funds.

Foreign currency accounts are moderately correlated with bank deposits (0.41). This indicates that customers who hold foreign currency accounts tend to have higher deposit levels, possibly reflecting higher income or international exposure.

### Income-Related Observations
Estimated income shows moderate positive correlations with several financial products, including superannuation savings (0.37), bank loans (0.33), and business lending (0.33). This implies that while income influences financial behaviour, it is not the only factor determining product usage.

### Weak Relationships
Superannuation savings exhibit relatively weak correlations with most other banking products. This suggests that retirement saving decisions are largely independent of day-to-day banking and credit behaviour.

Saving accounts show a weak correlation with business lending (0.31), indicating that personal saving habits do not strongly predict business borrowing behaviour.

### Overall Conclusion
The heatmap reveals that deposit-related products form a closely connected group, representing core banking activities. Lending products are moderately linked to deposits, suggesting that borrowers may still be financially stable and valuable to banks. Income plays a role in financial behaviour but does not fully explain product relationships, while superannuation savings operate as a distinct financial decision area.
