```python
import os
from dotenv import load_dotenv
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sqlalchemy import create_engine


load_dotenv()

HOST = os.getenv("HOST")
USER = os.getenv("USER")
PASSWORD = os.getenv("PASSWORD")
DATABASE = os.getenv("DB")
PORT = os.getenv("PORT")


try:
    engine = create_engine(f"mysql+mysqlconnector://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}")
    print(f"Engine created successfully for database: {DATABASE}")
     
except Exception as err:
    print(f"Error creating engine: {err}")
   
```

```python
query = """
SELECT
    *
FROM
    banking_db.banking;
"""
```

```python
df = pd.read_sql(query,engine)
```

### Exploratory Data Analyis (EDA)

```python
# calculate income band based on the Estimated Income and categorized as Low,Medium or High
bins = [0,100000,300000,float('inf')]
labels = ["Low","Medium","High"]

df["Income Band"] = pd.cut(df["Estimated Income"],bins=bins,labels=labels,right=False)
```

### count number of estimated income in each band

```python
df["Income Band"].value_counts()
```

### Visualize Income Band

```python
df["Income Band"].value_counts().plot(kind="bar")
```

#### Examine the distribution of categories in categorical columns

```python
category_cols = df[['BRId',
                    'GenderId',
                    'IAId',
                    'Amount of Credit Cards',
                    'Nationality',
                    'Occupation',
                    'Fee Structure',
                    'Loyalty Classification',
                    'Properties Owned',
                    'Risk Weighting',
                    'Income Band'
]]

for col in category_cols:
    print(f"Values for {col}")
    display(df[col].value_counts())
```

### visualize the data in categorical columns: Bivariate Analysis

```python
for i,col in enumerate(category_cols):
    plt.figure(i)
    sns.countplot(data=df,x=col,hue="GenderId")
```

### Univariate analysis

```python
for i,col in enumerate(category_cols):
    plt.figure(i)
    sns.countplot(data=df,x=col)
```
